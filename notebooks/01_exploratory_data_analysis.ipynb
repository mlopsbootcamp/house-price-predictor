# 🏠 House Data Exploration
This notebook performs initial data exploration on the `house_data.csv` dataset.

```python
# =============================================================
# NOTEBOOK 01 — EXPLORATORY DATA ANALYSIS (EDA)
# En MLOps, el EDA es el paso donde el Data Scientist/MLE
# entiende los datos ANTES de construir el modelo.
# El objetivo es detectar patrones, correlaciones y anomalías
# que guíen las decisiones de Feature Engineering y modelado.
# =============================================================

# Import required libraries
import pandas as pd          # Para manipular el dataset como tabla (DataFrame)
import numpy as np           # Para operaciones numéricas y matemáticas
import matplotlib.pyplot as plt  # Para crear gráficos base
import seaborn as sns        # Para gráficos estadísticos más visuales y expresivos
```

```python
# Load the dataset
# En MLOps, siempre cargamos desde data/raw/ — datos originales sin modificar.
# Esto garantiza reproducibilidad: cualquiera puede correr este notebook y obtener el mismo resultado.
df = pd.read_csv('../data/raw/house_data.csv')  # Lee el CSV y lo carga como DataFrame en memoria
                                                  # '../' sube un nivel desde notebooks/ a la raíz del proyecto

# ℹ️ Display basic information about the dataset
print(f"📐 Dataset shape (rows, columns): {df.shape}")  # f-string: imprime filas y columnas del dataset
                                                          # .shape devuelve una tupla (n_filas, n_columnas)
print("\n📊 Dataset Info:")   # '\n' agrega línea en blanco para separar visualmente la salida
print(df.info())              # .info() muestra: nombre de columna, tipo de dato y cantidad de valores no-nulos
                              # Muy útil para detectar columnas con datos faltantes o tipos incorrectos
```

```python
# ❓ Checking for missing values to identify data quality issues
# En MLOps, los valores faltantes son un problema serio:
# si en producción llega un registro con NaN, el modelo puede fallar o dar predicciones incorrectas.

print("\n🔍 Missing values per column:")  # Encabezado
print(df.isnull().sum())  # .isnull() crea una máscara booleana (True donde hay NaN)
                           # .sum() suma los True por columna = total de valores faltantes por columna
```

```python
# 📈 Descriptive statistics like mean, std deviation, min, max, etc.
# En MLOps, estas estadísticas sirven como 'baseline' para monitoreo en producción.
# Si el promedio de precios en producción difiere mucho del entrenamiento → data drift.

print("\n📋 Statistical Summary:")  # Encabezado
print(df.describe())  # .describe() calcula para cada columna numérica:
                      # count: total de valores no-nulos
                      # mean: promedio
                      # std: desviación estándar (qué tan dispersos están los valores)
                      # min / 25% / 50% / 75% / max: distribución de los datos
```

```python
# 💰 Visualizing the distribution of house prices
# En MLOps, entender la distribución del target (precio) es fundamental.
# Si está muy sesgada (skewed), puede ser necesario aplicar log-transform antes de entrenar.

plt.figure(figsize=(10, 6))       # Crea una figura de 10x6 pulgadas
sns.histplot(df['price'], kde=True)  # Histograma de la columna 'price'
                                      # kde=True: agrega curva de densidad (Kernel Density Estimate)
                                      # para ver la forma de la distribución más suavemente
plt.title('House Price Distribution')  # Título del gráfico
plt.xlabel('Price')                    # Etiqueta eje X
plt.ylabel('Frequency')               # Etiqueta eje Y (frecuencia = cuántas casas tienen ese precio)
plt.show()                            # Muestra el gráfico

print("📊 Plotted house price distribution.")  # Confirmación en consola
```

```python
# 🔗 Checking how features are correlated with each other
# En MLOps/ML, la correlación entre features y el target (precio) indica qué variables
# son más útiles para el modelo. Features muy correlacionadas entre sí pueden causar
# multicolinealidad, lo que afecta modelos lineales.

plt.figure(figsize=(12, 8))          # Figura más grande para que el heatmap sea legible
correlation_matrix = df.corr()       # .corr() calcula la correlación de Pearson entre todas las columnas numéricas
                                      # Valores cercanos a 1: correlación positiva fuerte
                                      # Valores cercanos a -1: correlación negativa fuerte
                                      # Valores cercanos a 0: sin correlación
sns.heatmap(correlation_matrix,      # Dibuja el mapa de calor con la matriz de correlación
            annot=True,              # annot=True: muestra el valor numérico en cada celda
            cmap='coolwarm',         # Paleta de colores: azul (negativo) → rojo (positivo)
            fmt='.2f')               # fmt='.2f': muestra los números con 2 decimales
plt.title('Feature Correlation Matrix')  # Título
plt.show()                               # Muestra el gráfico

print("📈 Displayed feature correlation heatmap.")  # Confirmación
```

```python
# 📏 Scatter plot to observe relationship between sqft and price
# En MLOps, este gráfico valida si existe una relación lineal entre una feature y el target.
# Si los puntos forman una nube sin patrón → la feature no es predictiva.
# Si forman una línea diagonal → hay correlación y la feature es útil para el modelo.

plt.figure(figsize=(10, 6))                    # Figura de 10x6
sns.scatterplot(x='sqft', y='price', data=df)  # Scatter plot: cada punto es una casa
                                                # x='sqft': eje X = metros cuadrados
                                                # y='price': eje Y = precio
                                                # data=df: usa el DataFrame df como fuente de datos
plt.title('Price vs Square Footage')           # Título
plt.xlabel('Square Footage')                   # Etiqueta eje X
plt.ylabel('Price')                            # Etiqueta eje Y
plt.show()                                     # Muestra el gráfico

print("📐 Analyzed relation between square footage and price.")  # Confirmación
```

```python
# 📍 Bar chart to see how many houses exist in each location
# En MLOps, entender la distribución de variables categóricas (como 'location') es importante.
# Si una categoría tiene muy pocos ejemplos, el modelo no aprenderá bien para esa categoría.
# Esto se llama 'class imbalance' o desbalance de clases.

plt.figure(figsize=(10, 6))                       # Figura de 10x6
sns.countplot(x='location', data=df)              # countplot: cuenta cuántas filas hay por cada valor único de 'location'
                                                   # Es equivalente a un histograma pero para variables categóricas
plt.title('Houses by Location')                   # Título
plt.xlabel('Location')                            # Etiqueta eje X
plt.ylabel('Count')                               # Etiqueta eje Y (cantidad de casas)
plt.xticks(rotation=45)                           # Rota las etiquetas del eje X 45 grados
                                                   # Necesario cuando los nombres de ubicación son largos y se solapan
plt.show()                                        # Muestra el gráfico

print("📍 Visualized distribution of houses by location.")  # Confirmación
```

```python
# 🛏️ Bar chart showing how average house price varies by bedroom count
# En MLOps, este análisis ayuda a entender si 'bedrooms' es una feature relevante.
# Si el precio promedio varía significativamente entre 1, 2, 3, 4 habitaciones → es una buena feature.

plt.figure(figsize=(10, 6))                         # Figura de 10x6
sns.barplot(x='bedrooms', y='price', data=df)       # barplot: para cada valor único de 'bedrooms',
                                                     # calcula el promedio de 'price' y lo muestra como barra
                                                     # También muestra el intervalo de confianza (barra de error)
plt.title('Average Price by Number of Bedrooms')    # Título
plt.xlabel('Number of Bedrooms')                    # Etiqueta eje X
plt.ylabel('Average Price')                         # Etiqueta eje Y
plt.show()                                          # Muestra el gráfico

print("🛏️ Compared average prices based on bedroom count.")  # Confirmación
```
